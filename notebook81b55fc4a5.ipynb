```python
import pandas as pd
import numpy as np

df_raw = pd.read_csv("/kaggle/input/datasets/aleenf/project/day15_real_dataset_large (1).csv")

def clean_data_project(df):
    df = df.copy()
    df["age"] = pd.to_numeric(df["age"], errors="coerce")
    df["income"] = pd.to_numeric(df["income"], errors="coerce")
    df["signup_time"] = pd.to_datetime(df["signup_time"], errors="coerce")
    df.loc[df["age"] <= 0, "age"] = np.nan
    df["age_missing"] = df["age"].isna().astype(int)
    df["income_missing"] = df["income"].isna().astype(int)
    df["age"].fillna(df["age"].median(), inplace=True)
    df["income"].fillna(df["income"].median(), inplace=True)
    df["income"] = df["income"].clip(upper=df["income"].quantile(0.99))
    df["age"] = df["age"].clip(lower=df["age"].quantile(0.01), upper=df["age"].quantile(0.99))
    df["city"] = df["city"].astype(str).str.strip().str.lower()
    df["signup_time"] = df["signup_time"].dt.tz_localize("UTC", ambiguous='NaT', nonexistent='NaT')
    return df

df_clean = clean_data_project(df_raw)
df_clean.to_csv("cleaned_dataset.csv", index=False)
df_clean.to_csv("cleaned_dataset.csv",index=False)
print(df_clean.info())
print(df_clean[["age","income"]].describe())
print(df_clean["city"].value_counts().head())
print(df_clean["signup_time"].dt.tz)
print(df_clean.isna().sum())
```
